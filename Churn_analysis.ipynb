```python
def conv(value):
    if value==1:
        return 'yes'
    else:
        return 'no'

df['SeniorCitizen']=df['SeniorCitizen'].apply(conv)
```

```python
# converted 0 and 1 values of senior Citizen to yes/no to make it easier to understand.
```

```python
ax=sns.countplot(x='Churn' ,data=df)
ax.bar_label(ax.containers[0])
plt.title('Count of Customer by Churn')
plt.show()
```

```python
plt.figure(figsize=(3,4))
gb= df.groupby('Churn').agg({'Churn':'count'})
gb
plt.pie(gb['Churn'], labels=gb.index, autopct= '%1.2f%%')
plt.title('Percentage of Churned Customers', fontsize=10)
plt.show()
```

```python
# from the given pie chart we can conclude that 26.54% of our customers have churned out.# now let's explore the reason behind it.
```

```python
plt.figure(figsize=(3,3))
sns.countplot(x='gender', data=df, hue='Churn')
plt.title('Churn by Gender')
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x='SeniorCitizen', data=df, hue='Churn')
plt.title('Churn by SeniorCitizen')
plt.show()
```

```python
plt.figure(figsize=(3,4))
ax=sns.countplot(x='SeniorCitizen', data=df)
ax.bar_label(ax.containers[0])
plt.title('Count of Customers by Senior Citizen')
plt.show()
```

```python
# Calculate percentages within each SeniorCitizen group
percentages = pd.crosstab(
    df['SeniorCitizen'],
    df['Churn'],
    normalize='index'
) * 100

# Plot stacked bars
ax = percentages.plot(
    kind='bar',
    stacked=True,
    figsize=(4, 4)
)

# Add percentage labels
for container in ax.containers:
    labels = [
        f'{bar.get_height():.1f}%' if bar.get_height() >= 5 else ''
        for bar in container
    ]
    ax.bar_label(container, labels=labels, label_type='center')

plt.title('Churn by SeniorCitizen (%)')
plt.xlabel('SeniorCitizen (0 = No, 1 = Yes)')
plt.ylabel('Percentage of Customers (%)')
plt.xticks(rotation=0)
plt.legend(title='Churn',bbox_to_anchor = (.9,.9))
plt.tight_layout()
plt.show()
```

# Comparatively a greater percentage of people in senior citizen category have churned out.

```python
columns = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

fig, axes = plt.subplots(3, 3, figsize=(18, 12))

axes = axes.flatten()

for i, col in enumerate(columns):
    sns.countplot(data=df, x=col, ax=axes[i], hue='Churn')
    axes[i].set_title(f'Count of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
    axes[i].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

#Customers without Online Security and Tech Support show noticeably higher churn, while customers with these services tend to be more loyal. Fiber-optic internet users have relatively high churn, and customers with month-to-month services may need further analysis.

#Overall, Online Security, Tech Support, and Internet Service type appear to be important factors to investigate for customer churn.
